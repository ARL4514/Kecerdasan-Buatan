JOBSHEET 3: PENGUKURAN STATISTIK DESKRIPTIF

LATIHAN

1. Central Tendency

A. Mendapatkan Nilai Mean (Rata-rata)

```python
import numpy as np

score = [90, 80, 70, 65, 110, 80, 83, 67, 94]
x = np.mean(score)

print(f"Data: {score}")
print(f"Nilai Mean: {x}")
```

B. Mendapatkan Nilai Median (Nilai Tengah)

```python
import numpy as np

score = [90, 80, 70, 65, 110, 80, 83, 67, 94]
x = np.median(score)

print(f"Data: {score}")
print(f"Nilai Median: {x}")
```

C. Mendapatkan Nilai Modus (Mode)

```python
from scipy import stats

score = [90, 80, 70, 65, 110, 80, 83, 67, 94]
x = stats.mode(score, keepdims=True)

print(f"Data: {score}")
print(f"Nilai Modus: {x.mode[0]}")
print(f"Jumlah Muncul: {x.count[0]}")
```

2. Spread (Penyebaran Data)

A. Mendapatkan Nilai Range (Rentang)

```python
import numpy as np

data = [2, 3, 7, 8, 9, 12, 15, 20, 50, 42]
minimum = np.min(data)
maximum = np.max(data)
x = maximum - minimum

print(f"Data: {data}")
print(f"Min: {minimum}, Max: {maximum}")
print(f"Nilai Range: {x}")
```

B. Mendapatkan Nilai Percentile

```python
import numpy as np

data = [90, 80, 70, 65, 110, 80, 83, 67, 94]
x = np.percentile(data, 90)  # Mencari persentil ke-90

print(f"Data: {data}")
print(f"Nilai Persentil ke-90: {x}")
```

C. Mendapatkan Nilai Quartile dan Interquartile Range (IQR)

```python
import numpy as np

data = [3, 4, 4, 5, 5, 6, 6, 7, 8, 8, 9, 10]
data.sort() # Memastikan data terurut

Q1 = np.median(data[:int(len(data)/2)])
Q3 = np.median(data[int(len(data)/2):])
IQR = Q3 - Q1

print(f"Data: {data}")
print(f"Q1: {Q1}, Q3: {Q3}, IQR: {IQR}")
```

D. Mendapatkan Nilai Variance dan Standar Deviasi

```python
import numpy as np

data = [86, 87, 88, 86, 87, 85, 86]
mean = np.mean(data)
var = np.var(data)
std = np.std(data)

print(f"Data: {data}")
print(f"Mean: {mean:.2f}")
print(f"Variance: {var:.2f}")
print(f"Standar Deviasi: {std:.2f}")
```

2. Percentile ke-25 dan Percentile ke-75

```python
import numpy as np

data = [9,10,11,6,8,7,7,5,4,5]

p25 = np.percentile(data, 25)
p75 = np.percentile(data, 75)

print("Percentile 25:", p25)
print("Percentile 75:", p75)
```

3. Analisis Data Nilai Ujian

```python
import numpy as np

# Data nilai ujian
data = [50, 54, 62, 50, 52, 59, 61, 63, 65, 10,
        53, 63, 65, 50, 59, 62, 50, 51, 57, 60,
        63, 65, 65, 53, 99]

print("Data awal:")
print(data)

# a. Mengurutkan data dari kecil ke besar
sorted_data = sorted(data)

print("\nA. Data setelah diurutkan:")
print(sorted_data)

# b. Menghitung Q1, Q3, dan IQR
Q1 = np.percentile(sorted_data, 25)
Q3 = np.percentile(sorted_data, 75)
IQR = Q3 - Q1

print("\nB. Quartile dan IQR:")
print("Q1 =", Q1)
print("Q3 =", Q3)
print("IQR =", IQR)

# c. Menghitung lower bound dan upper bound
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("\nC. Batas Outlier:")
print("Lower Bound =", lower_bound)
print("Upper Bound =", upper_bound)

# d. Menentukan nilai outlier
outliers = [x for x in sorted_data if x < lower_bound or x > upper_bound]

print("\nD. Nilai Outlier:")
print(outliers)
```

4. Standard Deviasi Berat Badan Mahasiswa

```python
import numpy as np

intervals = [
    (41,45,2),
    (46,50,3),
    (50,55,6),
    (56,60,4),
    (61,65,5),
    (66,70,2)
]

data = []

for a,b,f in intervals:
    midpoint = (a + b) / 2
    data += [midpoint] * f

std_dev = np.std(data)

print("Standard Deviasi:", std_dev)
```

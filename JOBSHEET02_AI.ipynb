3. Menggunakan Fungsi MATH

```python
import math

# Contoh penggunaan fungsi math
print("math.sin(0):", math.sin(0))                # Sinus
print("math.cos(0):", math.cos(0))                # Cosinus
print("math.tan(0):", math.tan(0))                # Tangen
print("math.log(10):", math.log(10))              # Logaritma natural
print("math.log10(10):", math.log10(10))          # Logaritma basis 10
print("math.degrees(math.pi):", math.degrees(math.pi)) # Konversi radian ke derajat
print("math.dist([0,0], [3,4]):", math.dist([0,0], [3,4])) # Jarak Euclidean
print("math.factorial(5):", math.factorial(5))    # Faktorial
print("math.fmod(10, 3):", math.fmod(10, 3))      # Sisa pembagian (float)
print("math.isnan(float('nan')):", math.isnan(float('nan'))) # Cek apakah NaN
print("math.gamma(5):", math.gamma(5))            # Fungsi Gamma
print("math.perm(5, 2):", math.perm(5, 2))        # Permutasi
```

4. Operasi Matriks dan vektor dengan python

a. Membuat Matriks di python

```python
import numpy as np
A = np.array([[80, 30, 25], [50, 20, 10]])
B = np.array([[30, 60, 15], [10, 30, 30]])
hasil = A + B
print("Hasil Penjumlahan Matriks A + B:")
print(hasil)
```

b. Melakukan operasi dot product

```python
import numpy as np
A = np.array([10, 20, 15])
B = np.array([16, 67, 35])
hasil = np.dot(A, B)
print("hasil: ", hasil)
```

c. Melakukan operasi cross product

```python
import numpy as np
A = np.array([10, 20, 15])
B = np.array([16, 67, 35])
Hasil = np.cross(A, B)
print("hasil: ", hasil)
```

TUGAS PRAKTIKUM

1. Perkalian Matriks

```python
import numpy as np

# Mendefinisikan matriks A, B, dan C
A = np.array([[3, 1], [2, 0]])
B = np.array([[2, 1], [1, -1]])
C = np.array([[-1, 1, 2], [2, 0, -1]])

# Perkalian Matriks A x B
hasil_AB = np.dot(A, B) # Menggunakan fungsi dot product sesuai modul [cite: 171]

# Perkalian Matriks B x C
hasil_BC = np.dot(B, C)

print("Hasil A x B:\n", hasil_AB)
print("Hasil B x C:\n", hasil_BC)
```

2. Operasi Faktorial

```python
import math

# Menghitung 6!
faktorial_6 = math.factorial(6) # Menggunakan fungsi math.factorial()

# Menghitung 8! / 5!
faktorial_gabungan = math.factorial(8) / math.factorial(5)

print("Nilai 6! =", faktorial_6)
print("Nilai 8! / 5! =", faktorial_gabungan)
```

3. Analisis Vektor 3D

```python
import numpy as np

# Mendefinisikan vektor dalam array 3D [cite: 169]
A = np.array([5, 3, 7])
B = np.array([12, -3, 1])

# Menghitung panjang vektor (Norm)
# Rumus: sqrt(sum(x^2))
panjang_A = np.linalg.norm(A)
panjang_B = np.linalg.norm(B)

# Perkalian Titik (Dot Product)
dot_AB = np.dot(A, B) # Sesuai contoh pada baris [cite: 171]

# Perkalian Silang (Cross Product)
cross_AB = np.cross(A, B) # Sesuai contoh pada baris [cite: 180]

print(f"Panjang Vektor A: {panjang_A:.2f}")
print(f"Panjang Vektor B: {panjang_B:.2f}")
print(f"Hasil A . B: {dot_AB}")
print(f"Hasil A x B: {cross_AB}")
```
